```python
try:
    x=int(input('Enter any number'))
    y=int(input('Enter any number'))
    if y!=0:
        result=x/y
        print(result)
    else:
        print('Y is zero')
except ValueError:
    print('Given input is not in correct format')
```

```python
# finally block
try:
    x=int(input('Enter any number'))
    y=int(input('Enter any number'))
    if y!=0:
        result=x/y
        print(result)
    else:
        print('Y is zero')
except ValueError:
    print('Given input is not in correct format')
finally:
    print('Finally block here ......')
```

```python
try:
    x=int(input('Enter any number'))
    y=int(input('Enter any number'))
    result=x/y
    print(result)
except (ZeroDivisionError,ValueError):
    print('Given input is not in correct format')
finally:
    print('Finally block here ......')
```

```python
try:
    x=int(input('Enter any number'))
    y=int(input('Enter any number'))
    result=x/y
    print(result)
except ZeroDivisionError as zd:
    print('Zero division error')
    print(zd)
except ValueError as ve:
    print('ValueError exception')
    print(ve)
except Exception as e:
    print('Exception super class')
    print(e)
finally:
    print('Finally block here ......')
```

```python
try:
    x=int(input('Enter any number'))
    y=int(input('Enter any number'))
    result=x/y
    print(result)

except Exception as e:
    print('Exception super class - exception raised')
    print(e)
finally:
    print('Finally block here ......')
```

```python
class MyException(Exception):
    def __str__(self):
        return super().__str__()
```

```python
try:
    raise MyException('I am raising my own exception')
except MyException as me:
    print('Custom exception')
    print(me)
```
