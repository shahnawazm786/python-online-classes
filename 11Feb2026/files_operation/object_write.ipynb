```python
import pickle
def write_object(data):
    with open("user.pk1","ab") as pk:
        pickle.dump(data,pk)
    pk.close()

```

```python
user={"username":'Admin003',"Password":"admin@345678","Login":True}
type(user)
write_object(user)
```

```python
def read_object():
    with open("user.pk1","rb") as pb:
        content=pickle.load(pb)
        print(content)
        pb.close()
        pb=None
```

```python
read_object()
```
