```python
f=None
print(f)
try:
    f=open("my_file.txt","r")
    print(f)
    content=f.read() # complete file read
    print(content)
except FileNotFoundError as fe:
    print('File not found')
    f=None
finally:
    f.close()
    f=None

print(f)
print(type(f))

```

```python
with open("my_file.txt","r") as fi:
    data=fi.read()
    print(data)
    fi.close()
    fi=None

print(fi)
print(type(fi))
    
```

```python
with open("my_file.txt","r") as fi:
    line=fi.readline()
    while True:
        print(line)
        line=fi.readline()
        if not line:
            break

    fi.close()
    fi=None
```

```python
with open("my_file.txt","r") as fi:
    line=fi.readline(6)
    while True:
        print(line)
        line=fi.readline(6)
        if not line:
            break

    fi.close()
    fi=None
```
